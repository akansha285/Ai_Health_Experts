```python
from pydantic import BaseModel
from typing import List

class HealthResponse(BaseModel):
    possible_conditions: List[str]
    precautions: List[str]
    severity: str
    advice: str
    emergency: bool
```

```python
from pydantic_ai import Agent

health_agent = Agent(
    "groq:llama3-70b-8192",
    output_type=HealthResponse,
    system_prompt="""
    You are an AI Health Assistant.

    Rules:
    - Never provide final diagnosis
    - Suggest only possible conditions
    - Always recommend doctor consultation
    - Detect emergencies:
        chest pain
        breathing issues
        severe bleeding
        unconsciousness
    - If emergency → severity HIGH
    - emergency = True
    - advice = Seek immediate medical help
    """
)
```

```python
def get_health_response(symptoms: str):
    result = health_agent.run_sync(
        f"Patient symptoms: {symptoms}"
    )
    return result.output
```

```python
import gradio as gr

def chat(symptoms):
    response = get_health_response(symptoms)

    return f"""
Possible Conditions: {', '.join(response.possible_conditions)}

Precautions: {', '.join(response.precautions)}

Severity: {response.severity}

Emergency: {response.emergency}

Advice: {response.advice}
"""

demo = gr.Interface(
    fn=chat,
    inputs="text",
    outputs="text",
    title="AI Health Assistant"
)

demo.launch()
```
